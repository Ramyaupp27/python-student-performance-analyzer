# Student Performance Analyzer

**Project level:** Beginner Python / Data Analytics practice  
**Language:** Python  
**Environment:** Jupyter Notebook

## Project overview

This project analyzes a small sample of student marks. It calculates total and average marks, identifies pass/fail status, finds each student's highest and lowest marks, demonstrates `map()` and `filter()`, and generates a performance report.

## Learning objectives

- Store related records in a Python dictionary.
- Create reusable functions with parameters and `return` statements.
- Use conditional statements to classify results.
- Use built-in functions such as `sum()`, `len()`, `max()`, and `min()`.
- Apply `lambda` with `map()` and `filter()`.
- Use a `for` loop to generate reports for multiple students.

## Pass/fail rule

- Average marks **greater than or equal to 50**: `PASS`
- Average marks **below 50**: `FAIL`

> The dataset is sample practice data. The `map()` example adds 5 marks for demonstration; it does not change the original student records.


## 1. Create the student dataset

The dictionary uses each student's name as a key and their list of subject marks as the value.

```python
students = {
    "Ramya": [85, 90, 78, 95, 88],
    "John": [45, 55, 60, 40, 50],
    "Priya": [92, 95, 89, 96, 98],
    "David": [35, 40, 45, 50, 38],
    "Sarah": [70, 75, 80, 85, 90]
}

students
```

## 2. Calculate total and average marks

These functions accept a student name, retrieve that student's marks from the dictionary, and return the calculated result.

```python
def calculate_total_marks(name):
    marks = students[name]
    return sum(marks)


def calculate_average_marks(name):
    marks = students[name]
    return sum(marks) / len(marks)


print(f"Ramya's total marks: {calculate_total_marks('Ramya')}")
print(f"Ramya's average marks: {calculate_average_marks('Ramya')}")
```

## 3. Check pass/fail status

The function uses the average-mark rule defined above and returns either `"PASS"` or `"FAIL"`.

```python
def check_student_result(name):
    average_mark = calculate_average_marks(name)

    if average_mark >= 50:
        return "PASS"
    else:
        return "FAIL"


for name in students:
    print(f"{name}: {check_student_result(name)}")
```

## 4. Find the highest and lowest marks

This function returns two values: the highest mark and the lowest mark. Python groups the returned values into a tuple, which can be unpacked into two variables.

```python
def analyze_marks(name):
    marks = students[name]
    highest_mark = max(marks)
    lowest_mark = min(marks)
    return highest_mark, lowest_mark


highest, lowest = analyze_marks("Ramya")
print(f"Ramya's highest mark: {highest}")
print(f"Ramya's lowest mark: {lowest}")
```

## 5. Transform and filter marks with `map()` and `filter()`

### Task A: Increase marks by 5

`map()` applies the lambda operation to every mark and returns an iterator. `list()` converts the results into a list.

### Task B: Select marks of 50 or above

`filter()` keeps only the marks that satisfy the condition. This filters individual subject marks; it is not the same as the student's overall pass/fail result.

```python
def increase_marks(name):
    marks = students[name]
    updated_marks = list(map(lambda mark: mark + 5, marks))
    return updated_marks


ramya_updated_marks = increase_marks("Ramya")
print(f"Ramya's original marks: {students['Ramya']}")
print(f"Ramya's marks after adding 5: {ramya_updated_marks}")


def filter_passed_marks(name):
    marks = students[name]
    passed_marks = list(filter(lambda mark: mark >= 50, marks))
    return passed_marks


john_passed_marks = filter_passed_marks("John")
print(f"John's original marks: {students['John']}")
print(f"John's subject marks of 50 or above: {john_passed_marks}")
```

## 6. Generate the final student performance report

The report function creates a report for one student. The `for` loop calls that function once for every student in the dictionary.

```python
def generate_student_report(name):
    total = calculate_total_marks(name)
    average = calculate_average_marks(name)
    result = check_student_result(name)
    highest, lowest = analyze_marks(name)

    print(f"Student: {name}")
    print(f"Total Marks: {total}")
    print(f"Average Marks: {average:.1f}")
    print(f"Result: {result}")
    print(f"Highest Marks: {highest}")
    print(f"Lowest Marks: {lowest}")


print("========== STUDENT PERFORMANCE REPORT ==========")
print()

for name in students:
    generate_student_report(name)
    print()

print("================================================")
```
